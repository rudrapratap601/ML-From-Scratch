```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from SimLinReg import SimLinReg
```

```python
df = pd.read_csv(r"C:\Users\rudra\OneDrive\Desktop\ML-From-Scratch\linear_regression\simple_linear_regression\placement_Sreg.csv")
df.head()
```

```python
X = df.iloc[:,0].values
Y = df.iloc[:,1].values
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state = 2)
```

```python
X_test
```

```python
Y_test
```

```python
lin_reg = SimLinReg()
```

```python
lin_reg.fit(X_train, Y_train)
```

```python
print(f"m = ",lin_reg.get_coefficient())
print(f"b = ",lin_reg.get_intercept())
```

```python
Y_pred = lin_reg.predict(X_test)
```

```python
r2 = lin_reg.r2_score(Y_test, Y_pred)

print(f"""Mean Absolute Error  = {round(lin_reg.mae(Y_test, Y_pred), 2)}
Mean Squared Error   = {round(lin_reg.mse(Y_test, Y_pred), 2)}
Root MSE             = {round(lin_reg.rmse(Y_test, Y_pred), 2)}
R2 Score             = {round(r2, 2)}
Adjusted R2 Score    = {round(lin_reg.adj_r2_score(X_test, r2), 2)}
""")
```
