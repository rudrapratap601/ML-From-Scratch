```python
import numpy as np

from sklearn.datasets import load_diabetes

from sklearn.model_selection import train_test_split
from GDLinReg import GDLinReg
```

```python
X, Y = load_diabetes(return_X_y = True)
```

```python
print(X.shape)
print(Y.shape)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, train_size = 0.2, random_state = 2)
```

```python
lin_reg = GDLinReg(
    epochs=10000,
    lr=0.1,
    tolerance=1e-7,
    scale=True
)

lin_reg.fit(X_train, Y_train)

print("Coefficients:", lin_reg.get_coef_())
print("Intercept:", lin_reg.get_intercept_())
print("Iterations:", lin_reg.n_iterations_)
print("Final Gradient Norm:", lin_reg.gradient_norm_)
```

```python
Y_pred = lin_reg.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

r2 = r2_score(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
n = X_test.shape[0]
k = X_test.shape[1]


print(f"""Mean Absolute Error  = {round(mean_absolute_error(Y_test, Y_pred), 2)}
Mean Squared Error   = {round(mse, 2)}
Root MSE             = {round(np.sqrt(mse), 2)}
R2 Score             = {round(r2, 2)}
Adjusted R2 Score    = {round(1 - ((1 - r2) * (n - 1)) / (n - k - 1), 2)}
""")
```
