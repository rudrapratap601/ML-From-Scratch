```python
import numpy as np

from MBGDRegressor import MBGDRegressor

from sklearn.model_selection import train_test_split

from sklearn.datasets import load_diabetes
```

```python
X, Y = load_diabetes(return_X_y = True)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
```

### **Using my MBGDRegressor Class**

```python
model = MBGDRegressor(batch_size=50, epochs=1000, scale=True)

model.fit(X_train, Y_train)

Y_pred = model.predict(X_test)

print("Coefficients:", model.get_coef_())
print("Intercept:", model.get_intercept_())
print("Epochs:", model.n_iterations_)
print("Final loss:", model.loss_)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

r2 = r2_score(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
n, k= X_test.shape


print(f"""Mean Absolute Error  = {round(mean_absolute_error(Y_test, Y_pred), 2)}
Mean Squared Error   = {round(mse, 2)}
Root MSE             = {round(np.sqrt(mse), 2)}
R2 Score             = {round(r2, 2)}
Adjusted R2 Score    = {round(1 - ((1 - r2) * (n - 1)) / (n - k - 1), 2)}
""")
```
